# 02 — Análise e recomendação: quando reforçar o pessoal no Algarve?

Calcula o indicador principal face à média mensal, gera o gráfico com eixo no zero e conclusão no título, e regista a recomendação e os limites.


```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("../dados/dormidas_algarve_limpo.csv")
MESES = [
    "janeiro", "fevereiro", "março", "abril", "maio", "junho",
    "julho", "agosto", "setembro", "outubro", "novembro", "dezembro",
]

por_mes = df.groupby("mes")["dormidas"].sum().sort_index()
media_mensal = por_mes.mean()
mes_pico = MESES[por_mes.idxmax() - 1]
vezes_media = por_mes.max() / media_mensal
acima_media = [MESES[m - 1] for m in por_mes[por_mes > media_mensal].index]
quota_verao_pct = por_mes.loc[[6, 7, 8, 9]].sum() / por_mes.sum() * 100

print(f"Total analisado: {len(df)} registos ({df['ano'].min()}–{df['ano'].max()})")
print(f"Mês de pico: {mes_pico} ({vezes_media:.1f}x a média mensal)")
print(f"Meses acima da média: {', '.join(acima_media)}")
print(f"Quota de junho a setembro: {quota_verao_pct:.1f}%")
```

```python
fig, ax = plt.subplots(figsize=(8.5, 4.2))
cores = ["#1d4ed8" if v > media_mensal else "#94a3b8" for v in por_mes.values]
ax.bar([MESES[m - 1][:3] for m in por_mes.index], por_mes.values / 1_000_000, color=cores)
ax.axhline(media_mensal / 1_000_000, linestyle="--", color="#334155", label="Média mensal")
ax.set_ylim(bottom=0)
vezes_pt = f"{vezes_media:.1f}".replace(".", ",")
ax.set_title(f"{mes_pico.capitalize()} tem {vezes_pt} vezes as dormidas de um mês médio no Algarve")
ax.set_xlabel("Mês")
ax.set_ylabel("Dormidas acumuladas 2022–2024 (milhões)")
ax.legend(frameon=False)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

Path("../figuras").mkdir(exist_ok=True)
fig.savefig("../figuras/dormidas_por_mes.png", dpi=150, bbox_inches="tight")
plt.close(fig)
print("Gráfico guardado em ../figuras/dormidas_por_mes.png")
```

## Recomendação e limites

- **Recomendação:** Reforçar a equipa de **maio a setembro** (todos os meses acima da média), com contratação máxima em **agosto**.
- **Limites:** Os dados mostram quando há mais procura (dormidas), não a margem líquida nem o preço médio por quarto; cobre apenas 2022–2024.
